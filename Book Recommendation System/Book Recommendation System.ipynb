```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
import plotly.express as px
import plotly.graph_objects as go
```

```python
books = pd.read_csv("books.csv", sep=";", encoding="latin-1", skip_blank_lines= True, on_bad_lines= 'skip')
```

```python
books.head()
```

```python
ratings = pd.read_csv('ratings.csv', encoding= 'latin-1', on_bad_lines= 'skip', sep = ';')
```

```python
ratings.head()
```

```python
data = pd.merge(books, ratings ,  on= 'ISBN')
```

```python
data = data[['ISBN', 'Book-Title', 'Book-Author', 'Book-Rating']]
```

```python
data.head()
```

```python
data.rename(columns= {"ISBN": "bookID", "Book-Title	": "title", "Book-Author": "authors", "Book-Rating": "average_rating"}, inplace= True)
```

```python
# Now let’s have a look at the column information
data.info()
```

```python
# Now let’s see the distribution of average ratings of all the books
fig = px.histogram(data, x='average_rating', 
                   nbins=30, 
                   title='Distribution of Average Ratings')
fig.update_xaxes(title_text='Average Rating')
fig.update_yaxes(title_text='Frequency')
fig.show()
```
